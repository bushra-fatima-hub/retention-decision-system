```python
import pandas as pd

customers = pd.read_csv("../data/customers_clean.csv")
print(customers.shape)
customers.head()
```

```python
churn_rate = customers["is_churn"].mean() * 100
print(round(churn_rate, 2), "% of customers left")
```

8.99% means about 9 out of every 100 customers left in March 2017. That’s roughly 87,000 of the 970,960 customers.

```python
customers.groupby("auto_renew")["is_churn"].mean() * 100
```

does cancelling before affect leaving?

```python
customers.groupby("ever_cancelled")["is_churn"].mean() * 100
```

does listening behaviour differ between stayers and leavers?

```python
customers.groupby("is_churn")[["active_days", "total_secs"]].mean()
```

does how long someone has been a customer matter?

```python
customers.groupby("is_churn")["tenure_days"].mean()
```

## What I found

- 8.99% of customers left in March 2017, so leavers are rare. Accuracy alone would mislead, so I will judge models by how many leavers they catch.
- Payment behaviour matters most: churn was 30.3% with auto-renew off vs 3.9% with it on.
- Customers who had cancelled churned at 60.3%, but I need to check the timing, because a cancellation in the same month may be leaking the answer.
- Listening differed only slightly (27.1 vs 30.4 hours a month), and tenure barely differed at all.
